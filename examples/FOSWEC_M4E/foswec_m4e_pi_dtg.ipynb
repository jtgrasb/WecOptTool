```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from matplotlib.patches import Polygon, Circle


def animate_foswec_m4e_sideview(
    wec_tdom_full,
    waves,
    interval=None,
    stride=2,
    repeat=True,
    figsize=(9, 4),
    xlim=None,
    zlim=None,
    slow_down_ani=1.0,
    n_wave_x=100,
    wave_color="royalblue",
    wave_alpha=0.95,
    body_alpha=0.9,
    show_waterline=True,
    show_time_text=True,
    show_wave_fill=False,
):
    """
    Lightweight side-view animation of the full FOSWEC M4E model with incident waves.

    Parameters
    ----------
    wec_tdom_full : xarray Dataset
        Time-domain output for one controller case.
    waves : WecOptTool waves object
        Wave definition used in the solve.
    interval : float or None
        Frame interval in ms. If None, computed from simulation dt and stride.
    stride : int
        Plot every `stride`-th frame to reduce animation size.
    repeat : bool
        Whether animation repeats.
    figsize : tuple
        Figure size.
    xlim, zlim : tuple or None
        Axes limits.
    slow_down_ani : float
        >1 slower, <1 faster.
    n_wave_x : int
        Number of x-points for the wave curve.
    """

    # -----------------------------
    # Geometry
    # -----------------------------
    flapThicknessBottom = 0.04
    flapThicknessTop = 0.1
    flapCenterDistanceApart = 1.44
    flapHeight = 0.58
    flapDraft = 0.59

    platformFrameLength = 1.44
    platformFrameThickness = 0.05
    platformFrameTopDepth = 0.53 + 0.1

    fullWidth = 1.63
    columnsRadius = 0.1
    columnsDraft = 0.53 + 0.1 + 0.4
    columnsXY = [flapCenterDistanceApart / 2, fullWidth / 2 - columnsRadius]

    daqLength = 0.72
    daqHeight = 0.2
    daqTopDepth = 0.53 + 0.1 + 0.05

    # -----------------------------
    # Time histories
    # -----------------------------
    time = np.asarray(wec_tdom_full["time"].values)[::stride]

    surge = np.asarray(
        wec_tdom_full["pos"].isel(realization=0, influenced_dof=0).values
    )[::stride]
    heave = np.asarray(
        wec_tdom_full["pos"].isel(realization=0, influenced_dof=1).values
    )[::stride]
    pitch = np.asarray(
        wec_tdom_full["pos"].isel(realization=0, influenced_dof=2).values
    )[::stride]
    flap1_pitch = np.asarray(
        wec_tdom_full["pos"].isel(realization=0, influenced_dof=5).values
    )[::stride]
    flap2_pitch = np.asarray(
        wec_tdom_full["pos"].isel(realization=0, influenced_dof=8).values
    )[::stride]

    if len(time) < 2:
        raise ValueError("Animation requires at least two time samples.")

    dt_sim = float(time[1] - time[0])
    if interval is None:
        interval = max(1, int(1000.0 * dt_sim * slow_down_ani))

    # -----------------------------
    # Helpers
    # -----------------------------
    def rot2(theta):
        c = np.cos(theta)
        s = np.sin(theta)
        return np.array([[c, -s],
                         [s,  c]])

    def transform_points(points_xz, theta, translation_xz):
        R = rot2(theta)
        return points_xz @ R.T + np.asarray(translation_xz)

    # -----------------------------
    # Local geometry
    # -----------------------------
    platform_local = np.array([
        [-platformFrameLength / 2, -platformFrameTopDepth - platformFrameThickness],
        [ platformFrameLength / 2, -platformFrameTopDepth - platformFrameThickness],
        [ platformFrameLength / 2, -platformFrameTopDepth],
        [-platformFrameLength / 2, -platformFrameTopDepth],
    ])

    daq_local = np.array([
        [-daqLength / 2, -daqTopDepth - daqHeight],
        [ daqLength / 2, -daqTopDepth - daqHeight],
        [ daqLength / 2, -daqTopDepth],
        [-daqLength / 2, -daqTopDepth],
    ])

    col_x_positions = [-columnsXY[0], columnsXY[0]]
    col_rects_local = []
    for xcol in col_x_positions:
        rect = np.array([
            [xcol - columnsRadius, -columnsDraft],
            [xcol + columnsRadius, -columnsDraft],
            [xcol + columnsRadius, 0.0],
            [xcol - columnsRadius, 0.0],
        ])
        col_rects_local.append(rect)

    flap_local = np.array([
        [-flapThicknessBottom / 2, 0.0],
        [ flapThicknessBottom / 2, 0.0],
        [ flapThicknessTop / 2, flapHeight],
        [-flapThicknessTop / 2, flapHeight],
    ])

    flap1_hinge_local = np.array([-flapCenterDistanceApart / 2, -flapDraft])
    flap2_hinge_local = np.array([ flapCenterDistanceApart / 2, -flapDraft])

    # -----------------------------
    # Limits
    # -----------------------------
    if xlim is None:
        xlim = (-2.3, 2.3)
    if zlim is None:
        zlim = (-1.45, 0.35)

    # -----------------------------
    # Incident wave field
    # -----------------------------
    spatial_x = np.linspace(xlim[0], xlim[1], n_wave_x)

    omega = np.asarray(waves.omega.values, dtype=float)
    wave_complex = np.squeeze(np.asarray(waves.values))

    if wave_complex.ndim != 1:
        raise ValueError(
            f"Expected squeezed wave amplitudes to be 1D over omega, got shape {wave_complex.shape}"
        )

    amp = np.abs(wave_complex)
    phase = np.angle(wave_complex)
    k = omega**2 / 9.81  # deep-water approximation

    wave_elevations = np.zeros((len(time), len(spatial_x)))
    for i, t in enumerate(time):
        wave_elevations[i, :] = np.sum(
            amp[:, None] * np.cos(
                omega[:, None] * t - k[:, None] * spatial_x[None, :] + phase[:, None]
            ),
            axis=0,
        )

    # -----------------------------
    # Figure
    # -----------------------------
    fig, ax = plt.subplots(figsize=figsize)
    ax.set_aspect("equal")
    ax.set_xlim(*xlim)
    ax.set_ylim(*zlim)
    ax.set_xlabel("x [m]")
    ax.set_ylabel("z [m]")
    ax.set_title("FOSWEC M4E Side View")
    ax.grid(True, alpha=0.25)

    if show_waterline:
        ax.axhline(0.0, color="deepskyblue", linestyle="--", linewidth=1.0, alpha=0.7)

    wave_line, = ax.plot([], [], color=wave_color, lw=1.8, alpha=wave_alpha, zorder=1)

    platform_patch = Polygon(
        platform_local, closed=True, facecolor="0.45", edgecolor="k", alpha=body_alpha, zorder=5
    )
    daq_patch = Polygon(
        daq_local, closed=True, facecolor="tab:orange", edgecolor="k", alpha=body_alpha, zorder=6
    )

    column_patches = [
        Polygon(rect, closed=True, facecolor="tab:gray", edgecolor="k", alpha=0.85, zorder=4)
        for rect in col_rects_local
    ]

    flap1_patch = Polygon(
        flap_local + flap1_hinge_local, closed=True,
        facecolor="tab:red", edgecolor="k", alpha=body_alpha, zorder=7
    )
    flap2_patch = Polygon(
        flap_local + flap2_hinge_local, closed=True,
        facecolor="tab:green", edgecolor="k", alpha=body_alpha, zorder=7
    )

    ax.add_patch(platform_patch)
    ax.add_patch(daq_patch)
    for cp in column_patches:
        ax.add_patch(cp)
    ax.add_patch(flap1_patch)
    ax.add_patch(flap2_patch)

    hinge1_marker = Circle((flap1_hinge_local[0], flap1_hinge_local[1]), 0.015, color="k", zorder=8)
    hinge2_marker = Circle((flap2_hinge_local[0], flap2_hinge_local[1]), 0.015, color="k", zorder=8)
    ax.add_patch(hinge1_marker)
    ax.add_patch(hinge2_marker)

    time_text = None
    if show_time_text:
        time_text = ax.text(
            0.02, 0.97, "",
            transform=ax.transAxes,
            ha="left", va="top",
            fontsize=9,
            bbox=dict(facecolor="white", alpha=0.75, edgecolor="none"),
        )

    def update(i):
        x0 = surge[i]
        z0 = heave[i]
        th0 = pitch[i]
        th1 = flap1_pitch[i]
        th2 = flap2_pitch[i]

        wave_line.set_data(spatial_x, wave_elevations[i, :])

        platform_patch.set_xy(transform_points(platform_local, th0, [x0, z0]))
        daq_patch.set_xy(transform_points(daq_local, th0, [x0, z0]))

        for rect_local, patch in zip(col_rects_local, column_patches):
            patch.set_xy(transform_points(rect_local, th0, [x0, z0]))

        hinge1_global = transform_points(flap1_hinge_local.reshape(1, 2), th0, [x0, z0])[0]
        hinge2_global = transform_points(flap2_hinge_local.reshape(1, 2), th0, [x0, z0])[0]

        flap1_patch.set_xy(transform_points(flap_local, th0 + th1, hinge1_global))
        flap2_patch.set_xy(transform_points(flap_local, th0 + th2, hinge2_global))

        hinge1_marker.center = hinge1_global
        hinge2_marker.center = hinge2_global

        if show_time_text:
            time_text.set_text(
                f"t = {time[i]:.2f} s\n"
                f"surge = {x0:.2f} m, heave = {z0:.2f} m\n"
                f"pitch = {np.degrees(th0):.1f} deg"
            )

        artists = [
            wave_line,
            platform_patch,
            daq_patch,
            *column_patches,
            flap1_patch,
            flap2_patch,
            hinge1_marker,
            hinge2_marker,
        ]
        if show_time_text:
            artists.append(time_text)
        return artists

    anim = FuncAnimation(
        fig,
        update,
        frames=len(time),
        interval=interval,
        blit=False,
        repeat=repeat,
        repeat_delay=0,
    )

    plt.close(fig)
    return anim
```

```python
import matplotlib as mpl
from IPython.display import HTML
mpl.rcParams["animation.embed_limit"] = 30
plt.rcParams["animation.html"] = "jshtml"


case_name = "PI_coupled"
case = results[case_name]

anim = animate_foswec_m4e_sideview(
    case["wec_tdom_full"],
    study.waves,
    stride=3,
    slow_down_ani=1.0,
    figsize=(8, 4),
    xlim=(-2.3, 2.3),
    zlim=(-1.45, 0.35),
    n_wave_x=80,
    show_wave_fill=False,
)
```
